```python
from IPython.display import HTML, display

html = r'''
<!DOCTYPE html>
<html lang="ca">

<head>

<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">

<title>🗺️ El mapa secret</title>

<style>

*{
    box-sizing:border-box;
}

body{
    margin:0;
    font-family:Arial,sans-serif;
    background:linear-gradient(135deg,#dff6ff,#fff4d6);
    color:#263238;
}

.game{
    max-width:1050px;
    margin:auto;
    padding:18px;
}

.header{
    background:white;
    border-radius:25px;
    padding:18px;
    text-align:center;
    box-shadow:0 5px 18px rgba(0,0,0,.12);
    margin-bottom:15px;
}

.header h1{
    margin:5px;
    font-size:30px;
}

.header p{
    color:#667085;
}

.stats{
    display:flex;
    justify-content:center;
    gap:10px;
    flex-wrap:wrap;
}

.stat{
    background:#f1f5f9;
    padding:9px 15px;
    border-radius:15px;
    font-weight:bold;
}


/* ================= MAP ================= */

.map{
    position:relative;
    min-height:580px;
    background:#9bd27d;
    border:8px solid white;
    border-radius:30px;
    overflow:hidden;
    box-shadow:0 8px 25px rgba(0,0,0,.15);
}

.map:before{
    content:"";
    position:absolute;
    width:140%;
    height:130px;
    background:#74c9df;
    transform:rotate(-10deg);
    left:-20%;
    top:210px;
    opacity:.8;
}

.map:after{
    content:"";
    position:absolute;
    width:130%;
    height:90px;
    background:#74c9df;
    transform:rotate(15deg);
    left:-10%;
    top:380px;
    opacity:.7;
}

.path{
    position:absolute;
    height:15px;
    background:#ddc184;
    border-radius:20px;
    z-index:2;
}

.p1{
    width:220px;
    left:150px;
    top:170px;
    transform:rotate(20deg);
}

.p2{
    width:220px;
    left:370px;
    top:245px;
    transform:rotate(-18deg);
}

.p3{
    width:220px;
    left:580px;
    top:170px;
    transform:rotate(35deg);
}

.p4{
    width:220px;
    left:640px;
    top:315px;
    transform:rotate(20deg);
}

.location{
    position:absolute;
    z-index:5;
    width:125px;
    height:110px;
    border:0;
    border-radius:22px;
    background:white;
    box-shadow:0 5px 14px rgba(0,0,0,.18);
    cursor:pointer;
    font-size:45px;
    transition:.2s;
}

.location:hover{
    transform:scale(1.08);
}

.location small{
    display:block;
    font-size:14px;
    color:#555;
    font-weight:bold;
}

.lock{
    opacity:.5;
    cursor:not-allowed;
}

.lock:hover{
    transform:none;
}

.l1{left:35px;top:65px;}
.l2{left:260px;top:170px;}
.l3{left:500px;top:70px;}
.l4{left:735px;top:180px;}
.l5{left:300px;top:380px;}
.l6{left:670px;top:410px;}


/* ================= QUESTION ================= */

.questionBox{
    display:none;
    background:white;
    border-radius:28px;
    padding:25px;
    margin-top:15px;
    box-shadow:0 7px 22px rgba(0,0,0,.15);
    text-align:center;
}

.questionType{
    display:inline-block;
    background:#eaf2ff;
    color:#3564a8;
    padding:7px 14px;
    border-radius:15px;
    font-size:14px;
    font-weight:bold;
}

.progress{
    height:12px;
    background:#e8e8e8;
    border-radius:20px;
    overflow:hidden;
    margin:15px 0;
}

.progressBar{
    height:100%;
    background:#59b66b;
    width:0%;
    transition:.3s;
}

.picture{
    width:180px;
    height:150px;
    background:#f5f7fa;
    border:4px solid #edf0f4;
    border-radius:25px;
    display:flex;
    align-items:center;
    justify-content:center;
    font-size:85px;
    margin:15px auto;
}

.hint{
    display:inline-block;
    background:#fff3c4;
    padding:8px 17px;
    border-radius:15px;
    font-size:17px;
    margin:5px;
}

.question{
    font-size:24px;
    font-weight:bold;
    margin:18px;
}

.sentence{
    font-size:25px;
    margin:20px;
    line-height:1.6;
}

.options{
    display:grid;
    grid-template-columns:1fr 1fr;
    gap:12px;
    max-width:700px;
    margin:20px auto;
}

.option{
    border:0;
    padding:15px;
    border-radius:17px;
    background:#edf3ff;
    font-size:19px;
    cursor:pointer;
    font-weight:bold;
}

.option:hover{
    background:#dce8ff;
}

.disabled{
    pointer-events:none;
}

.result{
    min-height:70px;
    border-radius:18px;
    padding:12px;
    font-size:18px;
    font-weight:bold;
    margin:10px;
}

.correct{
    background:#dcf7e3;
    color:#237a39;
}

.wrong{
    background:#ffe1e1;
    color:#a12c2c;
}

.answer{
    font-size:24px;
    margin:7px;
}

.nextBtn,
.backBtn,
.restartBtn{
    border:0;
    border-radius:17px;
    padding:13px 25px;
    font-size:17px;
    font-weight:bold;
    cursor:pointer;
    margin:5px;
}

.nextBtn{
    background:#5b8def;
    color:white;
}

.backBtn{
    background:#eeeeee;
}


/* ================= TREASURE ================= */

.treasure{
    display:none;
    background:white;
    border-radius:30px;
    padding:35px;
    text-align:center;
    margin-top:15px;
    box-shadow:0 8px 25px rgba(0,0,0,.18);
}

.chest{
    font-size:120px;
    animation:bounce 1.2s infinite alternate;
}

@keyframes bounce{
    from{transform:translateY(0);}
    to{transform:translateY(-12px);}
}

.treasure h2{
    font-size:32px;
}

.finalScore{
    font-size:26px;
    font-weight:bold;
}


/* MOBILE */

@media(max-width:700px){

    .map{
        min-height:720px;
    }

    .location{
        width:105px;
        height:95px;
        font-size:35px;
    }

    .l1{left:15px;top:45px;}
    .l2{left:185px;top:145px;}
    .l3{left:15px;top:270px;}
    .l4{left:190px;top:365px;}
    .l5{left:15px;top:510px;}
    .l6{left:190px;top:590px;}

    .path{
        display:none;
    }

    .options{
        grid-template-columns:1fr;
    }

}

</style>

</head>


<body>

<div class="game">

<div class="header">

<h1>🗺️ El mapa secret</h1>

<p>
Resol les preguntes i descobreix el tresor!
</p>

<div class="stats">

<div class="stat">
⭐ Punts: <span id="score">0</span>
</div>

<div class="stat">
🗺️ Zones: <span id="zones">0</span>/6
</div>

<div class="stat">
📝 Preguntes: <span id="done">0</span>/36
</div>

</div>

</div>


<!-- ================= MAP ================= -->

<div class="map" id="map">

<div class="path p1"></div>
<div class="path p2"></div>
<div class="path p3"></div>
<div class="path p4"></div>


<button class="location l1"
onclick="openZone(0)">

🏰

<small>🔍 Zona 1</small>

</button>


<button class="location l2 lock"
id="loc1"
onclick="openZone(1)">

🌉

<small>🔒 Zona 2</small>

</button>


<button class="location l3 lock"
id="loc2"
onclick="openZone(2)">

💡

<small>🔒 Zona 3</small>

</button>


<button class="location l4 lock"
id="loc3"
onclick="openZone(3)">

🗼

<small>🔒 Zona 4</small>

</button>


<button class="location l5 lock"
id="loc4"
onclick="openZone(4)">

🌳

<small>🔒 Zona 5</small>

</button>


<button class="location l6 lock"
id="loc5"
onclick="openZone(5)">

🏠

<small>🔒 Zona 6</small>

</button>

</div>


<!-- ================= QUESTION ================= -->

<div class="questionBox" id="questionBox">

<div>

<b id="zoneTitle">Zona 1</b>

·

<span id="qNumber">1</span>/6

</div>


<div class="progress">

<div class="progressBar" id="progressBar"></div>

</div>


<div class="questionType" id="questionType">
🖼️ Mira i tria
</div>


<div class="picture" id="picture">
🏰
</div>


<div class="hint" id="hint">
Pista
</div>


<div class="question" id="question">
Què és?
</div>


<div id="extra"></div>


<div class="options" id="options"></div>


<div id="result"></div>


<button
class="nextBtn"
id="nextBtn"
onclick="nextQuestion()"
style="display:none">

➡️ Següent

</button>


<button
class="backBtn"
onclick="backToMap()">

🗺️ Tornar al mapa

</button>

</div>


<!-- ================= TREASURE ================= -->

<div class="treasure" id="treasure">

<div class="chest">
🎁
</div>

<h2>
🎉 Has trobat el tresor!
</h2>

<p>
Has completat les 36 preguntes.
</p>

<div class="finalScore">

⭐ Puntuació:
<span id="finalScore">0</span>

</div>

<p id="finalMessage"></p>

<button
class="restartBtn"
onclick="restartGame()">

🔄 Tornar a jugar

</button>

</div>

</div>


<script>


/* =====================================================
   QUESTION BANK

   IMPORTANT:
   Each zone contains DIFFERENT question types.
   The same word is NOT used for every question.
   ===================================================== */


const zones = [

/* ================= ZONE 1 ================= */

{

questions:[

{
type:"🖼️ Mira i tria",
picture:"🏰",
hint:"城堡",
question:"Què és?",
options:[
"el castell",
"el pont",
"el far",
"l'arbre"
],
answer:"el castell"
},

{
type:"🇨🇳 Del xinès al català",
picture:"❓",
hint:"桥",
question:"Com es diu «桥» en català?",
options:[
"el castell",
"el pont",
"la torre",
"la casa"
],
answer:"el pont"
},

{
type:"🔤 Tria l'article",
picture:"🏠",
hint:"房子",
question:"Quina opció és correcta?",
sentence:"___ casa és gran.",
options:[
"El",
"La",
"L'",
"Els"
],
answer:"La"
},

{
type:"🧩 Completa",
picture:"🌳",
hint:"树",
question:"Completa la frase:",
sentence:"M'agrada aquest ____.",
options:[
"arbre",
"pont",
"castell",
"far"
],
answer:"arbre"
},

{
type:"❓ Comprèn",
picture:"🌉",
hint:"桥",
question:"Per on podem passar d'un costat a l'altre?",
options:[
"Pel pont",
"Pel castell",
"Pel far",
"Per la casa"
],
answer:"Pel pont"
},

{
type:"🔀 Ordena",
picture:"🏰",
hint:"城堡",
question:"Quina frase és correcta?",
options:[
"El castell és gran.",
"Castell el gran és.",
"És gran castell el.",
"Gran és el castell."
],
answer:"El castell és gran."
}

]

},


/* ================= ZONE 2 ================= */

{

questions:[

{
type:"🖼️ Mira i tria",
picture:"🌉",
hint:"桥",
question:"Què és?",
options:[
"el pont",
"el far",
"el castell",
"la torre"
],
answer:"el pont"
},

{
type:"🇨🇳 Del xinès al català",
picture:"❓",
hint:"灯塔",
question:"Com es diu «灯塔» en català?",
options:[
"la torre",
"el far",
"el pont",
"el castell"
],
answer:"el far"
},

{
type:"🔤 Tria l'article",
picture:"🗼",
hint:"塔",
question:"Quina opció és correcta?",
sentence:"___ torre és alta.",
options:[
"El",
"La",
"L'",
"Els"
],
answer:"La"
},

{
type:"🧩 Completa",
picture:"🏰",
hint:"城堡",
question:"Completa:",
sentence:"El ____ és molt antic.",
options:[
"castell",
"casa",
"torre",
"arbre"
],
answer:"castell"
},

{
type:"🧠 Quina no correspon?",
picture:"🌉",
hint:"Busca l'element diferent",
question:"Quina paraula NO és un lloc o construcció?",
options:[
"el pont",
"el castell",
"el far",
"l'arbre"
],
answer:"l'arbre"
},

{
type:"💬 Comprèn",
picture:"💡",
hint:"灯塔",
question:"On podem veure una llum per orientar els vaixells?",
options:[
"Al far",
"Al pont",
"A la casa",
"Al castell"
],
answer:"Al far"
}

]

},


/* ================= ZONE 3 ================= */

{

questions:[

{
type:"🖼️ Mira i tria",
picture:"💡",
hint:"灯塔",
question:"Què és?",
options:[
"el far",
"la torre",
"el pont",
"el castell"
],
answer:"el far"
},

{
type:"🇨🇳 Del xinès al català",
picture:"❓",
hint:"塔",
question:"Com es diu «塔» en català?",
options:[
"la torre",
"el far",
"el pont",
"la casa"
],
answer:"la torre"
},

{
type:"🔤 Tria l'article",
picture:"🌳",
hint:"树",
question:"Quina opció és correcta?",
sentence:"___ arbre és molt gran.",
options:[
"El",
"La",
"L'",
"Les"
],
answer:"L'"
},

{
type:"🧩 Completa",
picture:"🏠",
hint:"房子",
question:"Completa:",
sentence:"La meva ____ és petita.",
options:[
"casa",
"castell",
"pont",
"far"
],
answer:"casa"
},

{
type:"🔀 Ordena",
picture:"🌳",
hint:"树",
question:"Quina frase és correcta?",
options:[
"L'arbre és verd.",
"Arbre l'és verd.",
"És arbre l'verd.",
"Verd arbre és."
],
answer:"L'arbre és verd."
},

{
type:"🧠 Comprèn",
picture:"🗼",
hint:"塔",
question:"Quina és alta?",
options:[
"La torre",
"El pont",
"La casa",
"L'arbre"
],
answer:"La torre"
}

]

},


/* ================= ZONE 4 ================= */

{

questions:[

{
type:"🖼️ Mira i tria",
picture:"🗼",
hint:"塔",
question:"Què és?",
options:[
"la torre",
"el far",
"el pont",
"el castell"
],
answer:"la torre"
},

{
type:"🇨🇳 Del xinès al català",
picture:"❓",
hint:"房子",
question:"Com es diu «房子» en català?",
options:[
"la casa",
"el castell",
"la torre",
"el far"
],
answer:"la casa"
},

{
type:"🔤 Tria l'article",
picture:"🏰",
hint:"城堡",
question:"Completa:",
sentence:"___ castell és molt gran.",
options:[
"El",
"La",
"L'",
"Les"
],
answer:"El"
},

{
type:"🧩 Completa",
picture:"🌉",
hint:"桥",
question:"Completa:",
sentence:"Jo camino pel ____.",
options:[
"pont",
"casa",
"arbre",
"far"
],
answer:"pont"
},

{
type:"❓ Quina no correspon?",
picture:"🏠",
hint:"Busca la paraula diferent",
question:"Quina paraula és una planta?",
options:[
"l'arbre",
"el pont",
"la casa",
"el castell"
],
answer:"l'arbre"
},

{
type:"💬 Comprèn",
picture:"🏰",
hint:"城堡",
question:"On pot viure un rei en una història?",
options:[
"Al castell",
"Al pont",
"Al far",
"A l'arbre"
],
answer:"Al castell"
}

]

},


/* ================= ZONE 5 ================= */

{

questions:[

{
type:"🖼️ Mira i tria",
picture:"🌳",
hint:"树",
question:"Què és?",
options:[
"l'arbre",
"la torre",
"el far",
"el pont"
],
answer:"l'arbre"
},

{
type:"🇨🇳 Del xinès al català",
picture:"❓",
hint:"城堡",
question:"Com es diu «城堡» en català?",
options:[
"el castell",
"la torre",
"la casa",
"el pont"
],
answer:"el castell"
},

{
type:"🔤 Tria l'article",
picture:"💡",
hint:"灯塔",
question:"Completa:",
sentence:"___ far és a prop del mar.",
options:[
"El",
"La",
"L'",
"Les"
],
answer:"El"
},

{
type:"🧩 Completa",
picture:"🗼",
hint:"塔",
question:"Completa:",
sentence:"La ____ és molt alta.",
options:[
"torre",
"pont",
"castell",
"arbre"
],
answer:"torre"
},

{
type:"🔀 Ordena",
picture:"🌉",
hint:"桥",
question:"Quina frase és correcta?",
options:[
"El pont és llarg.",
"Pont el és llarg.",
"És el pont llarg.",
"Llarg pont és el."
],
answer:"El pont és llarg."
},

{
type:"🧠 Comprèn",
picture:"🏠",
hint:"房子",
question:"On vivim normalment?",
options:[
"A casa",
"Al pont",
"Al far",
"A la torre"
],
answer:"A casa"
}

]

},


/* ================= ZONE 6 ================= */

{

questions:[

{
type:"🖼️ Mira i tria",
picture:"🏠",
hint:"房子",
question:"Què és?",
options:[
"la casa",
"el castell",
"el pont",
"el far"
],
answer:"la casa"
},

{
type:"🇨🇳 Del xinès al català",
picture:"❓",
hint:"树",
question:"Com es diu «树» en català?",
options:[
"l'arbre",
"la torre",
"el pont",
"el castell"
],
answer:"l'arbre"
},

{
type:"🔤 Tria l'article",
picture:"🌉",
hint:"桥",
question:"Completa:",
sentence:"___ pont és llarg.",
options:[
"El",
"La",
"L'",
"Les"
],
answer:"El"
},

{
type:"🧩 Completa",
picture:"💡",
hint:"灯塔",
question:"Completa:",
sentence:"El ____ és a prop del mar.",
options:[
"far",
"casa",
"arbre",
"pont"
],
answer:"far"
},

{
type:"❓ Tria la frase correcta",
picture:"🏰",
hint:"城堡",
question:"Quina frase és correcta?",
options:[
"El castell és gran.",
"La castell és gran.",
"El castell són gran.",
"Castell el gran és."
],
answer:"El castell és gran."
},

{
type:"🏆 Repte final",
picture:"🗺️",
hint:"Pensa en tot el que has après",
question:"Quina frase és correcta?",
options:[
"El pont és al costat del castell.",
"La pont és al costat del castell.",
"El pont és al costat de la castell.",
"Pont és el costat del castell."
],
answer:"El pont és al costat del castell."
}

]

}

];


/* =====================================================
   GAME VARIABLES
   ===================================================== */

let score=0;

let completedZones=0;

let totalDone=0;

let currentZone=0;

let currentQuestion=0;

let answered=false;


/* =====================================================
   OPEN ZONE
   ===================================================== */

function openZone(index){

    if(index>completedZones) return;

    currentZone=index;

    currentQuestion=0;

    document.getElementById("map").style.display="none";

    document.getElementById("treasure").style.display="none";

    document.getElementById("questionBox").style.display="block";

    showQuestion();
}


/* =====================================================
   SHOW QUESTION
   ===================================================== */

function showQuestion(){

    answered=false;

    const q=
        zones[currentZone]
        .questions[currentQuestion];


    document.getElementById("zoneTitle").innerText=
        "🔎 Zona "+(currentZone+1);


    document.getElementById("qNumber").innerText=
        currentQuestion+1;


    document.getElementById("questionType").innerText=
        q.type;


    document.getElementById("picture").innerText=
        q.picture;


    document.getElementById("hint").innerText=
        q.hint;


    document.getElementById("question").innerText=
        q.question;


    document.getElementById("extra").innerHTML="";


    if(q.sentence){

        document.getElementById("extra").innerHTML=
            "<div class='sentence'>"+
            q.sentence+
            "</div>";

    }


    document.getElementById("progressBar").style.width=
        (currentQuestion/6*100)+"%";


    document.getElementById("result").innerHTML="";

    document.getElementById("nextBtn").style.display="none";


    const box=
        document.getElementById("options");


    box.innerHTML="";


    /* RANDOM ANSWER ORDER */

    let options=[...q.options];

    options.sort(()=>Math.random()-0.5);


    options.forEach(option=>{

        const btn=
            document.createElement("button");

        btn.className="option";

        btn.innerText=option;

        btn.onclick=
            ()=>checkAnswer(option,btn);

        box.appendChild(btn);

    });

}


/* =====================================================
   CHECK ANSWER
   ===================================================== */

function checkAnswer(selected,button){

    if(answered) return;

    answered=true;

    const q=
        zones[currentZone]
        .questions[currentQuestion];


    document
    .querySelectorAll(".option")
    .forEach(b=>b.classList.add("disabled"));


    totalDone++;

    document.getElementById("done").innerText=
        totalDone;


    const result=
        document.getElementById("result");


    if(selected===q.answer){

        score+=10;

        button.style.background="#c8efd1";

        result.className="result correct";

        result.innerHTML=
            "✅ Molt bé!"+
            "<div class='answer'>"+
            q.answer+
            "</div>";


        speak(q.answer);

    }

    else{

        score-=2;

        button.style.background="#ffd1d1";

        result.className="result wrong";

        result.innerHTML=
            "❌ No és correcte."+
            "<div class='answer'>"+
            "Resposta: "+q.answer+
            "</div>";


        speak(q.answer);

    }


    document.getElementById("score").innerText=
        score;


    document.getElementById("nextBtn").style.display=
        "inline-block";

}


/* =====================================================
   NEXT
   ===================================================== */

function nextQuestion(){

    currentQuestion++;


    if(currentQuestion>=6){

        completeZone();

    }

    else{

        showQuestion();

    }

}


/* =====================================================
   COMPLETE ZONE
   ===================================================== */

function completeZone(){

    completedZones++;

    document.getElementById("zones").innerText=
        completedZones;


    const loc=
        document.getElementById(
            "loc"+(completedZones-1)
        );


    if(loc){

        loc.querySelector("small").innerText=
            "✅ Zona "+completedZones;

    }


    /* UNLOCK NEXT */

    if(completedZones<6){

        const next=
            document.getElementById(
                "loc"+completedZones
            );

        next.classList.remove("lock");

        next.querySelector("small").innerText=
            "🔍 Zona "+(completedZones+1);

    }


    document.getElementById("questionBox").style.display=
        "none";


    if(completedZones===6){

        showTreasure();

    }

    else{

        document.getElementById("map").style.display=
            "block";

    }

}


/* =====================================================
   BACK TO MAP
   ===================================================== */

function backToMap(){

    document.getElementById("questionBox").style.display=
        "none";

    document.getElementById("map").style.display=
        "block";

}


/* =====================================================
   TREASURE
   ===================================================== */

function showTreasure(){

    document.getElementById("map").style.display=
        "none";

    document.getElementById("questionBox").style.display=
        "none";

    document.getElementById("treasure").style.display=
        "block";


    document.getElementById("finalScore").innerText=
        score;


    let message="";


    if(score>=330){

        message=
        "🏆 Fantàstic! Ets un súper detectiu del català!";

    }

    else if(score>=270){

        message=
        "🌟 Molt bé! Has fet una gran feina!";

    }

    else if(score>=200){

        message=
        "👏 Molt bé! Continua practicant!";

    }

    else{

        message=
        "💪 Ho has intentat molt bé! Tornem a jugar!";

    }


    document.getElementById("finalMessage").innerText=
        message;

}


/* =====================================================
   SPEECH
   ===================================================== */

function speak(text){

    if("speechSynthesis" in window){

        window.speechSynthesis.cancel();

        const u=
            new SpeechSynthesisUtterance(text);

        u.lang="ca-ES";

        u.rate=.8;

        u.pitch=1;

        window.speechSynthesis.speak(u);

    }

}


/* =====================================================
   RESTART
   ===================================================== */

function restartGame(){

    location.reload();

}

</script>

</body>

</html>
'''

display(HTML(html))
```
